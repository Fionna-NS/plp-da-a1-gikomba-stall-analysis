```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100, 2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}
```

## Task 1: Basic sales statistics
I use Python’s built-in sum, len, max and min, then list.index()to find which day the best and worst sales happened.



```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest = max(daily_sales)
lowest = min(daily_sales)

best_pos = daily_sales.index(highest)
worst_pos = daily_sales.index(lowest)

best_day = days[best_pos]
best_week = best_pos // 7 + 1      # positions 0-6 are week 1, 7-13 are week 2

worst_day = days[worst_pos]       #same idea using the worst_pos
worst_week = worst_pos // 7 + 1   #same idea using the worst_pos

print(f"Total sales: {total_sales:,}")
print(f"Average sales: {average_sales:,.2f}")
print(f"Highest sales: {highest:,} on {best_day} in week {best_week}")
print(f"Lowest sales: {lowest:,} on {worst_day} in week {worst_week}")
```

## Task 2: Weekly totals and growth
I slice the list into two weeks, sum each slice, and return both totals as a tuple.

```python
def week_totals(sales):
    """Return a tuple (week 1 total, week 2 total) from a list of 14 daily sales."""
    week1 = sum(sales[0:7])
    week2 = sum(sales[7:14])
    return week1, week2

week1_total, week2_total = week_totals(daily_sales)

growth_pct = (week2_total - week1_total) / week1_total * 100

print(f"Week 1 total: {week1_total:,}")
print(f"Week 2 total: {week2_total:,}")
print(f"Growth: {growth_pct:,.2f}%")
```

## Task 3: Classifying days
I compare each day’s sales to the average with a function, then count the labels in a dictionary.

```python
def classify_day(amount, average):
    """Return 'Strong', 'Normal' or 'Weak' based on how amount compares to average."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

day_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    label = classify_day(amount, average_sales)
    day_counts[label] += 1

print(f"Strong days: {day_counts['Strong']}")
print(f"Normal days: {day_counts['Normal']}")
print(f"Weak days: {day_counts['Weak']}")
```

```python
for day, amount in zip(days, daily_sales):
    print(day, amount, classify_day(amount, average_sales))
```

##Task 4: Weekday vs weekend sales
I split the sales into two lists using the day names, average each list, and compare them.

```python
weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Sat", "Sun"]:
        weekend_sales.append(amount)
    else:
        weekday_sales.append(amount)

weekday_avg = sum(weekday_sales) / len(weekday_sales)
weekend_avg = sum(weekend_sales) / len(weekend_sales)  # same idea for the weekend list

difference_pct = (weekend_avg - weekday_avg) / weekday_avg * 100

print(f"Weekday average: {weekday_avg:,.2f}")
print(f"Weekend average: {weekend_avg:,.2f}")
print(f"Weekend sales are {difference_pct:.2f}% higher than weekday sales")
```

## Task 5: Restock list and gross margins
I loop through the stock dictionary with .items() to find low-stock products and to calculate each product’s margin.

```python
def products_to_reorder(stock):
    """Return a list of product names whose quantity is at or below the reorder level."""
    low_stock = []
    for name, info in stock.items():
        if info["qty"] <= info["reorder"]:
            low_stock.append(name)
    return low_stock

reorder_list = products_to_reorder(stock)
print("Products to reorder:", reorder_list)

for name, info in stock.items():
    margin = (info["price"] - info["cost"]) / info["price"] * 100
    print(f"{name:<10} margin: {margin:.1f}%")
```

## Task 6: Stall Performance Report
I reuse the results from Tasks 1-5 and print them with f-strings, using width and alignment so the columns line up.

```python
def report_row(label, value):
    """Print one report row: label left-aligned, value right-aligned, 45 characters wide."""
    print(f"{label:<22}{value:>23}")

print("=" * 45)
print("STALL PERFORMANCE REPORT".center(45))
print("Wanjiku's Fresh Produce Stall, Gikomba Market".center(45))
print("=" * 45)


print("\nSALES SUMMARY (14 days)")
print("-" * 45)
report_row("Total sales", f"KES {total_sales:,}")
report_row("Average daily sales", f"KES {average_sales:,.2f}")
report_row("Best day", f"KES {highest:,} ({best_day}, week {best_week})")
report_row("Worst day", f"KES {lowest:,} ({worst_day}, week {worst_week})")

print("\nWEEKLY TREND")
print("-" * 45)
report_row("Week 1 total", f"KES {week1_total:,}")
report_row("Week 2 total", f"KES {week2_total:,}")
report_row("Week-on-week growth", f"{growth_pct:.2f}%")

print("\nWEEKDAY VS WEEKEND")
print("-" * 45)
report_row("Weekday average", f"KES {weekday_avg:,.2f}")
report_row("Weekend average", f"KES {weekend_avg:,.2f}")
report_row("Weekend uplift", f"+{difference_pct:.2f}%")

print("\nDAY TYPES")
print("-" * 45)
report_row("Strong days", f"{day_counts['Strong']} days")
report_row("Normal days", f"{day_counts['Normal']} days")
report_row("Weak days", f"{day_counts['Weak']} days")

print("\nSTOCK AND MARGINS")
print("-" * 45)
print(f"{'Product':<10} {'Qty':>5} {'Reorder':>8} {'Margin':>8}")
for name, info in stock.items():
    margin = (info['price'] - info['cost']) / info['price'] * 100
    print(f"{name:<10} {info['qty']:>5} {info['reorder']:>8} {margin:>7.1f}%")

print(f"\nRestock now: {', '.join(reorder_list)}")
print("=" * 45)
```
